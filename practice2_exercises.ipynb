## Quick review — what is Merge Sort?

Merge sort is a **recursive** sorting algorithm that follows three steps:

1. **Divide** — split the list into two halves.
2. **Sort** — recursively sort each half (using merge sort again).
3. **Merge** — combine the two sorted halves into one sorted list.

The recursion stops (the **base case**) when a sublist has just one element — a single element is already "sorted."

**Complexity:**
- Time: **O(N log N)** — O(log N) recursion levels × O(N) work at each level to merge.
- Space: **O(N)** — we need temporary storage for the sublists during merging.

## The setup: `mergeSortHelp` (already given)

You do **not** need to write this function — read it and make sure you understand what it does. It's the "Divide + recursive Sort" part of the algorithm.

- **Inputs:** the list `L`, the leftmost index `first`, and the rightmost index `last` of the region we're currently sorting.
- **Base case:** if `first == last`, the region has just one element — nothing to do.
- **Recursive case:** find the middle, sort the two halves, then merge them back together.


````
def mergeSortHelp(L: list, first: int, last: int) -> None:
    if first == last:               # Base case: single-element region
        return
    else:                            # Recursive case
        # Step 1: Divide it into two sublists
        mid = first + (last - first) // 2
        # Step 2: Sort the two sublists (recursively)
        mergeSortHelp(L, first, mid)
        mergeSortHelp(L, mid + 1, last)
        # Step 3: Merge the two 'sorted' sublists
        merge(L, first, mid, last)
````

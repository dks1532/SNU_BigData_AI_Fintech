# Exercise 1 — Sum of Positive Integers

> Write a function `P1` that takes a list of integers as input, adds only the positive integers in the list, and returns the sum.


## Step 1 — Understand the problem

Let's break the problem down before touching any code.

| Item | Detail |
|---|---|
| **Input** | A list of integers, e.g. `[1, 2, -3]` |
| **Output** | A single integer — the sum of only the **positive** numbers |
| **Positive means** | Strictly greater than `0` (so `0` is **not** positive) |
| **Edge case 1** | List has no positive numbers → return `0` |
| **Edge case 2** | List is empty → return `0` |

**Trace through the first example by hand:**

```
Input:  [1, 2, -3]
Step 1: look at 1  → positive → running sum = 0 + 1 = 1
Step 2: look at 2  → positive → running sum = 1 + 2 = 3
Step 3: look at -3 → NOT positive → skip → running sum stays 3
Output: 3 ✓
```


## Step 2 — Plan the algorithm (in English)

1. Start with a running total `pos_sum = 0`.
2. Go through each number in the list one at a time.
3. If that number is greater than `0`, add it to `pos_sum`.
4. Otherwise, ignore it.
5. After the loop finishes, return `pos_sum`.

Notice that **starting `pos_sum` at `0` automatically handles both edge cases** — if we never find a positive number (or the list is empty), we just return the initial `0`. Neat!


## Step 3 — Beginner-friendly solution (for-loop + if)

This is the most straightforward translation of the plan above. It uses only concepts we've already covered: variables, `for` loops, `if` statements, and `return`.


````python
from typing import List

def P1(numbers: List[int]) -> int:
    """
    Returns the sum of positive integers in the given list.

    [Input]  numbers (List[int]): A list of integers.
    [Output] pos_sum (int): The sum of positive integers in the list.
    [Constraints] Return 0 if there are no positive integers in the list.
    """
    pos_sum = 0                 # 1. Start the running total at 0

    for n in numbers:           # 2. Look at each number one at a time
        if n > 0:               # 3. Is it positive?
            pos_sum = pos_sum + n   # 4. If yes, add it to the total
        # If not, we do nothing (skip it)

    return pos_sum              # 5. After the loop, return the total
````

### Let's test it against every example on the slide:

````python
print(P1([1, 2, -3]))     # Expected: 3
print(P1([-1, 0]))        # Expected: 0  (0 is NOT positive!)
print(P1([-5, -10, -3]))  # Expected: 0  (no positives at all)
print(P1([]))             # Expected: 0  (empty list)
````

> ⚠️ **Common mistake to watch for:** using `n >= 0` instead of `n > 0`. That would incorrectly count `0` as a positive number. Try changing it and see what `P1([-1, 0])` returns — it still gives `0` here because `0` adds nothing, but semantically the condition is wrong and it will bite you on other problems.


## Step 4 — More Pythonic alternative #1: `+=` operator

The line `pos_sum = pos_sum + n` is so common that Python has a shortcut for it: `pos_sum += n`. It means exactly the same thing. Get used to reading and writing it.


````python
def P1_v2(numbers: List[int]) -> int:
    pos_sum = 0
    for n in numbers:
        if n > 0:
            pos_sum += n       # ← same as pos_sum = pos_sum + n
    return pos_sum

print(P1_v2([1, 2, -3]))       # 3
print(P1_v2([-5, -10, -3]))    # 0
````

## Step 5 — More Pythonic alternative #2: `sum()` + generator expression

Python has a built-in `sum()` function that adds up all the numbers in an iterable. We can combine it with a **generator expression** to filter for positives on the fly — the whole function becomes a single line.


````python
def P1_v3(numbers: List[int]) -> int:
    return sum(n for n in numbers if n > 0)

print(P1_v3([1, 2, -3]))       # 3
print(P1_v3([-1, 0]))          # 0
print(P1_v3([-5, -10, -3]))    # 0
print(P1_v3([]))               # 0
````

**Why does the empty case still work?** `sum()` of an empty sequence is `0` by default. No special handling needed. This is a nice example of how Python's built-ins are designed to "do the right thing" for edge cases.

**Which version should you use?**
- **Learning phase:** Use version 1 (explicit for-loop). It makes the logic visible.
- **Real code:** Version 3 is what most Python programmers would write — it's short, readable, and idiomatic.

# Exercise 2 — Find Duplicates

> Take a list of integers as input argument, and return a set of numbers that appear two or more times.


## Step 1 — Understand the problem

| Item | Detail |
|---|---|
| **Input** | A list of integers, e.g. `[1, 1, 2, 3, 3, 3]` |
| **Output** | A **set** of integers — the ones that appear 2 or more times |
| **Why a set?** | Each duplicate value should appear **once** in the output, and order does not matter |
| **Edge case** | No duplicates → return an empty set `set()` |

**Trace through Example 2 by hand:**

```
Input:  [1, 1, 2, 3, 3, 3]
Count each number:
  1 appears 2 times → duplicate ✓
  2 appears 1 time  → not a duplicate ✗
  3 appears 3 times → duplicate ✓
Output: {1, 3} ✓
```

> ⚠️ Careful: the output is `{1, 3}`, **not** `{1, 3, 3, 3}`. Even though `3` appears three times, it goes into the set only once — that's what a set is for.


## Step 2 — Plan the algorithm (in English)

The problem has two natural sub-steps:

1. **Count** how many times each number appears in the list.
2. **Collect** the numbers whose count is `>= 2` into a set.

To count things, a **dictionary** is the perfect tool: keys are the numbers we see, values are how many times we've seen them.


## Step 3 — Beginner-friendly solution (dictionary counter)

We'll build a counting dictionary manually to see how counting with a dict works.


````python
from typing import List, Set

def P2(numbers: List[int]) -> Set[int]:
    """
    Returns a set of integers that appear two or more times in the input list.
    """
    # --- Phase 1: count how many times each number appears ---
    counts = {}                     # empty dictionary: number -> count

    for n in numbers:
        if n in counts:             # have we seen this number before?
            counts[n] = counts[n] + 1   # yes: increment its count
        else:
            counts[n] = 1           # no: start its count at 1

    # --- Phase 2: collect numbers with count >= 2 ---
    duplicates = set()              # start with an empty set

    for number, count in counts.items():
        if count >= 2:
            duplicates.add(number)

    return duplicates
````

### Test it:

````python
print(P2([1, 2, 3, 1]))         # Expected: {1}
print(P2([1, 1, 2, 3, 3, 3]))   # Expected: {1, 3}
print(P2([1, 2, 3, 4, 5]))      # Expected: set()
````

**Let's peek inside `counts` for the second example to make sure our intuition matches the code:**


````python
# Peek at the intermediate `counts` dictionary
numbers = [1, 1, 2, 3, 3, 3]
counts = {}
for n in numbers:
    if n in counts:
        counts[n] += 1
    else:
        counts[n] = 1

print('counts =', counts)   # {1: 2, 2: 1, 3: 3}
# Then we keep only keys whose value >= 2  →  {1, 3}
````

## Step 4 — More Pythonic alternative #1: `dict.get()`

The `if n in counts / else` pattern is so common when counting that Python has a cleaner form using `dict.get(key, default)`. It returns `counts[n]` if `n` exists, and the default value `0` otherwise — so we can always "add 1" without checking first.


````python
def P2_v2(numbers: List[int]) -> Set[int]:
    counts = {}
    for n in numbers:
        counts[n] = counts.get(n, 0) + 1   # ← one line replaces the if/else

    return {num for num, c in counts.items() if c >= 2}   # set comprehension

print(P2_v2([1, 2, 3, 1]))         # {1}
print(P2_v2([1, 1, 2, 3, 3, 3]))   # {1, 3}
print(P2_v2([1, 2, 3, 4, 5]))      # set()
````

The `{num for num, c in counts.items() if c >= 2}` part is a **set comprehension** — it's like a list comprehension but with `{}` instead of `[]`, so the result is a set. Read it as: *"the set of `num` for each `(num, c)` pair in counts, keeping only those where `c >= 2`."*
